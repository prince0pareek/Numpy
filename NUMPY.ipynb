### Shortcut method to create different types numpy array

```python
# zeros array creation - always create floating point
a = np.zeros(5)
print(a)
```

```python
#ones array creation - always create floating
a = np.zeros((4,3))
print(a)
```

```python
a = np.ones(4)
print(a)
```

```python
# any number array creation
a = np.full(10,4) # (part,value) # it create integer value
print(a)
```

```python
a = np.full((5,4),2) # ((Row,Colums),value)
print(a)
```
